```python
import pandas as pd
```

```python
# Load data into the dataframe
df = pd.read_csv("../climate_data.csv")
```

```python
# Check the first 5 rows
df.head()
```

```python
# Rename the columns to include units
df = df.rename(columns={
    'high_temp': 'high_temp_F',
    'low_temp': 'low_temp_F',
    'mean_temp': 'mean_temp_F',
    'precipitation': 'precipitation_in',
    'humidity': 'humidity_percent',
    'dew_point': 'dew_point_F',
    'wind': 'wind_mph',
    'pressure': 'pressure_Hg',
    'visibility': 'visibility_mi'
})
```

```python
# Clean temperatures and dew point (remove ' °F')
temp_cols = ['high_temp_F', 'low_temp_F', 'mean_temp_F', 'dew_point_F']
for col in temp_cols:
    df[col] = df[col].str.replace(' °F', '', regex=False).astype(float)

# Clean precipitation (remove quotes)
df['precipitation_in'] = df['precipitation_in'].str.replace('"', '', regex=False).astype(float)

# Clean humidity (remove '%')
df['humidity_percent'] = df['humidity_percent'].str.replace('%', '', regex=False).astype(float)

# Clean wind and visibility (remove ' mph', ' mi')
df['wind_mph'] = df['wind_mph'].str.replace(' mph', '', regex=False).astype(float)
df['visibility_mi'] = df['visibility_mi'].str.replace(' mi', '', regex=False).astype(float)

# Clean pressure (remove ' "Hg')
df['pressure_Hg'] = df['pressure_Hg'].str.replace(' "Hg', '', regex=False).astype(float)
```
